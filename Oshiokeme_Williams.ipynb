# Class 1

```python
import statistics
ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]

# This shows how many people are in the set
num_people = len(ages)
print(f"Total people: {num_people}")

#this shows the total ages of everyone in the set
total_age = sum(ages)
print(f"Total of all ages: {total_age}")

#this shows the mean age to one decimal place
mean_age = total_age / num_people
print(f"Mean age: {mean_age:.1f}")

#this function shows the highest and lowest age
youngest = min(ages)
oldest = max(ages)
print(f"Youngest age: {youngest}, Oldest age: {oldest}")

median_age = statistics.median(ages)
print(f"Median age: {median_age}")
```

```python
import pandas as pd

#This loads the file
df = pd.read_excel("50_startups_messy.xlsx")

# This gives the amount or number of rows and columns
print("Shape:", df.shape)

# This reports the missing counts per column
print("Missing counts:\n", df.isnull().sum())

# This reports duplicate count
print("Duplicate count:", df.duplicated().sum())
```

```python

df_clean = df.drop_duplicates().copy()

# 2. Standardise State column ('new york' -> 'New York')
df_clean["State"] = df_clean["State"].str.strip().str.title()

df_clean["Marketing Spend"] = pd.to_numeric(
    df_clean["Marketing Spend"], errors="coerce"
)
```

```python
srs_sample = df_clean.sample(n=20, random_state=7)

strat_sample = df_clean.groupby("State", group_keys=False).apply(
    lambda x: x.sample(frac=0.20, random_state=7)
)
```

The Simple Random Sample is much closer to the true population average for both R&D and Marketing spend, and this is because there is a bigger sample size and the groups were alresady equal
